```python

# Create a working copy of the original dataset
df_clean = df_raw.copy()

print("Working copy created successfully!")
print("Original dataset shape:", df_raw.shape)
print("Working dataset shape:", df_clean.shape)
```

```python

# Select numerical columns
numerical_columns = df_clean.select_dtypes(include=np.number).columns

# Calculate possible outliers using the IQR method
for column in numerical_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print(f"{column}: {len(outliers)} possible outliers")
```

```python

# Check categorical columns
categorical_columns = df_clean.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}")
    print(df_clean[column].value_counts(dropna=False))
```

```python

from sklearn.model_selection import train_test_split

# First split: 70% training and 30% temporary data
train_split, temp_split = train_test_split(
    df_clean,
    test_size=0.30,
    random_state=42
)

# Second split: divide temporary data equally
validation_split, test_split = train_test_split(
    temp_split,
    test_size=0.50,
    random_state=42
)

# Display the sizes
print("Training dataset:", train_split.shape)
print("Validation dataset:", validation_split.shape)
print("Testing dataset:", test_split.shape)
```

```python

# Columns containing missing values
columns_to_fill = [
    "rainfall_mm",
    "soil_ph",
    "pest_incidence_index"
]

# Calculate medians from training data only
training_medians = train_split[columns_to_fill].median()

print("Training data medians:")
print(training_medians)
```

```python

# Create separate copies
train_clean = train_split.copy()
validation_clean = validation_split.copy()
test_clean = test_split.copy()

# Fill missing values using training medians
train_clean[columns_to_fill] = train_clean[columns_to_fill].fillna(training_medians)

validation_clean[columns_to_fill] = validation_clean[columns_to_fill].fillna(training_medians)

test_clean[columns_to_fill] = test_clean[columns_to_fill].fillna(training_medians)

print("Missing values handled successfully!")
```

```python

print("Training missing values:")
print(train_clean[columns_to_fill].isnull().sum())

print("\nValidation missing values:")
print(validation_clean[columns_to_fill].isnull().sum())

print("\nTesting missing values:")
print(test_clean[columns_to_fill].isnull().sum())
```

```python

# Function to create new features
def create_features(data):

    data = data.copy()

    # 1. Total chemical fertilizer
    data["chemical_fertilizer_kg_ha"] = (
        data["npk_kg_ha"] + data["urea_kg_ha"]
    )

    # 2. Convert manure from tonnes to kilograms
    data["manure_kg_ha"] = data["manure_t_ha"] * 1000

    # 3. Calculate total fertilizer input
    data["total_fertilizer_input_kg_ha"] = (
        data["chemical_fertilizer_kg_ha"] +
        data["manure_kg_ha"]
    )

    # 4. Express plant density in thousands
    data["plant_density_per_1000"] = (
        data["plant_density_per_ha"] / 1000
    )

    # 5. Classify soil pH
    data["soil_ph_category"] = np.select(
        [
            data["soil_ph"] < 5.5,
            data["soil_ph"].between(5.5, 7.5),
            data["soil_ph"] > 7.5
        ],
        ["Acidic", "Near-neutral", "Alkaline"],
        default="Unknown"
    )

    # 6. Classify altitude
    data["altitude_zone"] = np.select(
        [
            data["altitude_m"] < 1500,
            data["altitude_m"].between(1500, 2000),
            data["altitude_m"] > 2000
        ],
        ["Lowland", "Midland", "Highland"],
        default="Unknown"
    )

    return data


# Apply feature engineering to all three datasets
train_clean = create_features(train_clean)
validation_clean = create_features(validation_clean)
test_clean = create_features(test_clean)

print("Feature engineering completed successfully!")
print("Training shape:", train_clean.shape)
print("Validation shape:", validation_clean.shape)
print("Testing shape:", test_clean.shape)
```
